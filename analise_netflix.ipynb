```python
#   IMPORTANDO BIBLIOTECAS  #
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#   IMPORTANDO A DATABASE   #
df = pd.read_csv('netflix_titles.csv')
df.head()
```

```python
#    Extraindo informações  #
df.info()
```

```python
#   Procurando valores nulos    #
df.isnull().sum()
```

```python
#   Descobrindo se existem mais Filmes ou mais Séries   #
df.value_counts(df['type'])
```

```python
#   Visualizando essa informação    #
sns.countplot(data=df, x='type')
plt.title('Filmes VS Series')
plt.xlabel('TIPO')
plt.ylabel('QUANTIDADE')
plt.show()
```

```python
#   Quais paises produzem mais filmes?  #
df.value_counts(df['country'])
```

```python
#   Visualizando o top 5 produtores   #
top_cinco= pd.DataFrame( df['country'].value_counts().head(5))
top_cinco.plot(xlabel=('Pais'))
plt.show()
```

```python
# Em quais anos a Netflix mais lançou títulos? Será que esse crescimento foi explosivo nos últimos anos? #
```

```python
filmes_anos_2000 = df.loc[df['release_year'] >= 2000, 'release_year'].value_counts().sort_index()
filmes_anos_2000.plot(xlabel=('Ano de Lançamento'))
plt.show()
```

```python
# Descobrindo a faixa etária do catálogo #
top_classificacao = pd.DataFrame(df['rating'].value_counts().head(10).reset_index())
top_classificacao.plot.bar(x='rating', xlabel = ('Classificação Indicativa'), ylabel = ('Produções'))
plt.show()
```
